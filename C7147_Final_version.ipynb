## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
sequence = ""
reference = "ATGC"

def validate_read(sequence, reference):
    """Validates a sequence against a reference, converts the sequence to upper case then checks if it is the same length
    and contains only ACGT. Returns the upper case sequence, valid which is only true if the length and characters are
    matched and a list of errors if there are any."""
    errors = []
  #adds a wrong length string to 'errors'
    sequence=sequence.upper()
    if len(reference) != len(sequence):
      errors.append("wrong_length")
  #adds invalid characters string to 'errors'
    for base in sequence:
      if base not in "ATCG":
        errors.append("invalid_characters")
        break
    if errors == []:
      valid = True
    else:
      valid = False
    return {"sequence": sequence, "valid": valid, "errors": errors}

#print(validate_read(sequence, reference))
# # # # # # # # # # # # # # # # # # # #


````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
#sequence = "GTGA"
#reference = "ATTC"

def find_mutations(sequence, reference):
  """Compares a read sequence to a read sequence of the same length base by base. If there is a difference it adds a 1d list
   to a dictionary with the position of the difference and both of the bases, otherwise the list would be empty"""

  change_list = []
  for base in range(len(sequence)):
    if sequence[base] != reference[base]:
      change_list.append({"position":base + 1, "reference_base": reference[base],"read_base": sequence[base]})
  return change_list


#find_mutations(sequence, reference)
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
#Write your code here

def analysed_reads(reads, reference):

  """Validates every read against the reference and find mutations from the valid ones.
   Returns a new list of dictionaries, one per read in the original order, with read_id,
    sample, group, the upper case sequence then status (either valid or invalid), errors
     and then mutations, any invalid reads have mutations = []. The original dictionaries
      are not changed"""


  gene_list = []
#for all valid genes find mutations and add them to the 'mutations' variable
  for gene in reads:
    sequence_result = validate_read(gene["sequence"], reference)

    if sequence_result["valid"]:
      status = "valid"
      possible_mutations = find_mutations(sequence_result["sequence"], reference)

    else:
      status = "invalid"
      possible_mutations = []

#create each dictionary
    gene_list.append({"read_id": gene["read_id"],
                     "sample": gene["sample"],
                     "group": gene["group"],
                     "sequence": sequence_result["sequence"],
                     "status": status,
                     "errors": sequence_result["errors"],
                     "mutations": possible_mutations})

  return gene_list
````

### 💻 Enter your code here

````python
# Write your code here.

def summarise_samples(analysed_reads):

  """Summarises any differences and mutations for each sample and then lists them according to which base was changed,
   taken from analyse_reads. Counts valid and invalid reads per sample, only valid reads then calculate reads with at
   least one change, the total number of change events and the percentage of valid reads with a change, none if there
   arent any valid reads. Also the count of each substitution (from base to base), the count at each position and the
   top three positions, returns a dictionary with ones one nested dictionary per sample."""

#loops through every read and keeps a running tally for each sample
  groups = {}
  for gene in analysed_reads:
      individual_sample = gene["sample"]
      if individual_sample not in groups:
        groups[individual_sample] ={"group": gene["group"],
                                    "valid_reads": 0,
                                    "invalid_reads": 0,
                                    "mismatch_reads": 0,
                                    "mismatch_percentage": None,
                                    "mismatch_events": 0,
                                    "substitution_counts": {},
                                    "position_counts": {},
                                    "top_positions": []}
      stored = groups[individual_sample]
      if gene["status"] == "valid":
        stored["valid_reads"] +=1
      else:
        stored["invalid_reads"] += 1

      if gene["mutations"]:
        stored["mismatch_reads"] +=1

#tallies every mismatch per sample, then works out the final stats once all reads are processed
      for instance in gene["mutations"]:
        stored["mismatch_events"] += 1
        change = instance["reference_base"]+ " to " + instance["read_base"]
        stored["substitution_counts"][change] = stored["substitution_counts"].get(change, 0) + 1
        position = instance["position"]
        stored["position_counts"][position] = stored["position_counts"].get(position, 0) + 1
  for individual_sample in groups:
    stored = groups[individual_sample]
    if stored["valid_reads"] > 0:
      stored["mismatch_percentage"] = stored["mismatch_reads"] / stored["valid_reads"] * 100
    positions = sorted(stored["position_counts"], key=lambda temp: (-stored["position_counts"][temp], temp))
    stored["top_positions"] = positions[:3]
  print_summary(groups)
  return groups

#print the output in a easy to read table
def print_summary(groups):

    """Prints the tables for each group, either "treatment" or "control" followed by each sample
     in the group and what each base change resulted in and how often the mutation occured."""

    new_list = {}
    for section in groups:
        segment = groups[section]["group"]
        if segment not in new_list:
            new_list[segment] = []
        new_list[segment].append(section)

    for segment in sorted(new_list):
        samples = sorted(new_list[segment])

        print("GROUP:", segment)
        print("=" * 14)
        print("")
        print("Summarised results:")
        print("===================")
        print("")
        print(f"{'Sample':<8}{'Valid':>12}{'Invalid':>12}{'% mismatch':>18}{'Events':>11}   Top positions")
        print("-" *77)
        print("")
        for section in samples:
            groups_part = groups[section]
            if groups_part["mismatch_percentage"] is None:
                percentage = "N/A"
            else:
                percentage = f"{groups_part['mismatch_percentage']:.1f}%"
            top_ranked = ", ".join(str(pos_temp) for pos_temp in groups_part["top_positions"])
            if top_ranked == "":
                top_ranked = "-"
            print(f"{section:<8}{groups_part['valid_reads']:>11}{groups_part['invalid_reads']:>10}{percentage:>18}{groups_part['mismatch_events']:>14}   {top_ranked:>14}")
        print("")
        print("")
        print("Changed base mutations count and % per 100 valid reads:")
        print("=======================================================")
        print("")

      #print base mutations ( A to C ect.)

        all_subs = set()
        for section in samples:
            all_subs.update(groups[section]["substitution_counts"])

        if len(all_subs) == 0:
            print("No substitutions found in data")
        else:
            print(f"{'  Change':<9}" + "".join(f"{section:>18}" for section in samples))
            length_of_heading = (len(f"{'  Change':<9}" + "".join(f"{section:>18}" for section in samples)))
            print("-" * length_of_heading)
            for sub in sorted(all_subs):
                individual_row = f"{sub.replace('to', ' into '):<10}"
                for section in samples:
                    groups_part = groups[section]
                    count = groups_part["substitution_counts"].get(sub, 0)
                    if groups_part["valid_reads"] == 0:
                        cell = f"{count} (N/A)"
                    else:
                        cell = f"{count} ({count / groups_part['valid_reads'] * 100:.1f}%)"
                    individual_row += f"{cell:>18}"
                print(individual_row)
        print("")
        print()






````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `sample_summary` is a dictionary in the format returned by the `summarise_samples` function
- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

Reuse helper functions from Step 4 where appropriate to reduce code complexity and repetition.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

### 📋 Return format

Return a list with one dictionary per selected position:

[
    {

        "treated_mean": 2.0,

    }

Means are sample mismatch percentages; `difference` is **treated mean − control mean**, in percentage points. Use `None` for results that cannot be calculated.

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
# Write your code here.

#function that finds all the samples that belong to one group and returns their names as a list
def get_samples_in_group(sample_summary, group_name):
    """Return the names of all samples that are in 1 group"""
    sample_names = []
    for sample_name in sample_summary:
        if sample_summary[sample_name]["group"] == group_name:
            sample_names.append(sample_name)
    return sample_names

#picks the positions with the highest mismatch counts, and if two positions have the same count, the earlier one comes first
def select_top_positions(position_counts, number_of_positions=3):
    """Return the positions with the highest counts, earlier position first if they are the same"""
    sorted_positions = sorted(position_counts, key=lambda position: (-position_counts[position], position))
    return sorted_positions[:number_of_positions]

# = = = = = = = = = = = = = = = = =

def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
    """Compare mismatch percentages between 2 groups at the 3 most mutated positions. The mismatch counts are aded together from
     all samples in both group, with the top 3 positions selected (earlier winning in case of a tie). In each position the
     mismatch percentages are calculated for each sample (mismatch/valid reads*100), and then the control and treated percentages
      are compared with calculate_statistics(2 sided, welch t-test). Samples without valid reads are excluded and a message displayed.
       The function returns a list with one dictionary per selected top position, and prints each on one line. """

    control_samples = get_samples_in_group(sample_summary, control_group)
    treated_samples = get_samples_in_group(sample_summary, treated_group)

# sorts every sample into usable control, usable treated, or excluded
    usable_control_samples= []
    usable_treated_samples =[]
    for sample_name in control_samples+ treated_samples:
        if sample_summary[sample_name]["valid_reads"] == 0:
            print(f"Excluded {sample_name}: no valid reads so no percentage to be calculated")
        elif sample_name in control_samples:
            usable_control_samples.append(sample_name)
        else:
            usable_treated_samples.append(sample_name)

#adds up the mismatch counts at each position across every sample in both groups, giving one combined tally
    combined_position_counts = {}
    for sample_name in control_samples+ treated_samples:
        sample_position_counts = sample_summary[sample_name]["position_counts"]
        for position in sample_position_counts:
            combined_position_counts[position] = (combined_position_counts.get(position, 0)+ sample_position_counts[position])

    top_positions = select_top_positions(combined_position_counts)

# builds the list of control mismatch percentages for one position, with one number per control sample
    comparison_results= []
    for position in top_positions:
        control_percentages = []
        for sample_name in usable_control_samples:
            sample_data = sample_summary[sample_name]
            mismatch_count = sample_data["position_counts"].get(position,0)
            control_percentages.append(mismatch_count/ sample_data["valid_reads"]* 100)

        treated_percentages = []
        for sample_name in usable_treated_samples:
            sample_data = sample_summary[sample_name]
            mismatch_count = sample_data["position_counts"].get(position, 0)
            treated_percentages.append(mismatch_count/ sample_data["valid_reads"]* 100)

#calculate percentages if there are values, if not returns empty
        if len(control_percentages) == 0:
          control_mean = None
        else:
          control_mean = sum(control_percentages) / len(control_percentages)
        if len(treated_percentages) == 0:
            treated_mean= None
        else:
            treated_mean= sum(treated_percentages) / len(treated_percentages)


        if control_mean is None or treated_mean is None:
            difference= None
        else:
            difference = treated_mean - control_mean

#works out the difference between the groups, then runs the statistical test, and stores everything for this position as one dictionary
        statistical_result= calculate_statistics(control_percentages, treated_percentages)
        comparison_results.append({"position": position,
                                   "control_n": len(control_percentages),
                                   "treated_n": len(treated_percentages),
                                   "control_mean": control_mean,
                                   "treated_mean": treated_mean,
                                   "difference": difference,
                                   "statistic": statistical_result["statistic"],
                                   "p_value": statistical_result["p_value"]})



#Prints one line per position and changes whats printed based on the data
        if difference is None:
            print(f"Position {position}: comparison not possible as the group has no usable samples)")
        elif statistical_result["p_value"] is None:
            print(f"Position {position}: control {control_mean:.2f}% (n={len(control_percentages)}), "
                  f"treated {treated_mean:.2f}% (n={len(treated_percentages)}), "
                  f"difference {difference:+.2f} points, cannot run Welch's t-test")
        else:
            print(f"Position {position}: control {control_mean:.2f}% (n={len(control_percentages)}), "
                  f"treated {treated_mean:.2f}% (n={len(treated_percentages)}), "
                  f"difference {difference:+.2f} points, "
                  f"Welch's t = {statistical_result['statistic']:.2f}, "
                  f"p = {statistical_result['p_value']:.4f}")

    return comparison_results
````
